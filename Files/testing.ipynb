```python

alone_player = None
players_left = [1, 2, 3]

if alone_player in players_left:
    print(f'Player {alone_player} is alone in the game')
elif alone_player is None:
    print('There is no alone player')
else:
    print(f'Player {alone_player} is not in the game')
```

```python
"""  """

lista_jogadores_por_jogo = [(1, 3), (2, 4),(5, 4),(1, 2)]

for i, (j1, j2) in enumerate(lista_jogadores_por_jogo):
    print(f'Jogo {i + 1}: Jogador {j1} vs Jogador {j2}')
    if j1 >= j2:
        print(f'Jogador {j1} ganhou de {j2}')
    else:
        print(f'Jogador {j2} ganhou de {j1}')
```
